```python
import pandas as pd
import os

# Folder containing rainfall data files
folder_path = r"C:\Users\nikhi\Desktop\ALL_NRSVIC\Rainfall_2018-24"
rainfall_dfs = []

for filename in os.listdir(folder_path):
    print("🔍 Found file:", filename)
    file_path = os.path.join(folder_path, filename)
    
    try:
        # Read CSV files
        if filename.lower().endswith('.csv'):
            df = pd.read_csv(file_path)
        # Read Excel files
        elif filename.lower().endswith(('.xlsx', '.xls')):
            df = pd.read_excel(file_path)
        else:
            print(f"⚠️ Skipping unsupported file type: {filename}")
            continue

        # Parse 'Date' safely
        df['Date'] = pd.to_datetime(df['Date'], errors='coerce')

        # Drop rows with invalid dates
        before = len(df)
        df.dropna(subset=['Date'], inplace=True)
        after = len(df)

        # Remove time portion from date
        df['Date'] = df['Date'].dt.date

        rainfall_dfs.append(df)
        print(f"✅ Loaded: {filename} ({before - after} bad rows dropped)")
        print("   📅 Sample Date:", df['Date'].iloc[0])

    except Exception as e:
        print(f"❌ Error reading {filename}: {e}")

# Combine all DataFrames
combined_rainfall_df = pd.concat(rainfall_dfs, ignore_index=True)

# Sort by State → District → Date
combined_rainfall_df.sort_values(by=['State', 'District', 'Date'], inplace=True)
combined_rainfall_df.reset_index(drop=True, inplace=True)

# Save to final CSV
output_path = 'combined_rainfall_data_2018_2024.csv'
combined_rainfall_df.to_csv(output_path, index=False)

# Confirm years present
combined_rainfall_df['Year'] = pd.to_datetime(combined_rainfall_df['Date']).dt.year
print(f"\n✅ Combined rainfall data saved to: {output_path}")
print("📆 Years present in combined data:", sorted(combined_rainfall_df['Year'].unique()))
```

```python
import pandas as pd
from sklearn.linear_model import LinearRegression
import numpy as np

# Load original seasonal rainfall data (1901–2021)
file_path = r'C:\Users\nikhi\Desktop\ALL_NRSVIC\TEMP_ANNUAL_SEASONAL_MEAN.csv'  # 🔁 update if needed
seasonal_df = pd.read_csv(file_path)

# Columns to predict
columns_to_predict = ['ANNUAL', 'JAN-FEB', 'MAR-MAY', 'JUN-SEP', 'OCT-DEC']

# Ensure all values are numeric (coerce errors to NaN)
seasonal_df[columns_to_predict] = seasonal_df[columns_to_predict].apply(pd.to_numeric, errors='coerce')

# Drop rows with missing values
cleaned_df = seasonal_df.dropna(subset=columns_to_predict + ['YEAR'])

# Train linear models and predict for 2022–2024
years = cleaned_df['YEAR'].values.reshape(-1, 1)
predictions = []

for col in columns_to_predict:
    model = LinearRegression()
    model.fit(years, cleaned_df[col])
    future_years = np.array([[2022], [2023], [2024]])
    predicted = model.predict(future_years)
    predictions.append(predicted)

# Create prediction rows
predicted_rows = pd.DataFrame({
    'YEAR': [2022, 2023, 2024],
    'ANNUAL': predictions[0],
    'JAN-FEB': predictions[1],
    'MAR-MAY': predictions[2],
    'JUN-SEP': predictions[3],
    'OCT-DEC': predictions[4],
})

# Append to original (even if it had missing years)
final_df = pd.concat([seasonal_df, predicted_rows], ignore_index=True)

# Sort by year (just in case)
final_df.sort_values(by='YEAR', inplace=True)
final_df.reset_index(drop=True, inplace=True)

# ✅ Save back to the same file
final_df.to_csv(file_path, index=False)
print(f"✅ Predicted 2022–2024 data saved into: {file_path}")
```

```python
import pandas as pd

# ✅ Load merged daily rainfall + soil data with correct date parsing
daily_df = pd.read_csv(
    "merged_rainfall_soil_data.csv",
    parse_dates=['Date'],
    dayfirst=True  # Ensures correct DD-MM-YYYY interpretation
)

# ✅ Extract 'Year' column if not already present
if 'Year' not in daily_df.columns:
    daily_df['Year'] = daily_df['Date'].dt.year

# ✅ Load temperature data (extended to 2024)
temp_df = pd.read_csv(r'C:\Users\nikhi\Desktop\ALL_NRSVIC\TEMP_ANNUAL_SEASONAL_MEAN.csv')

# ✅ Rename temperature columns for clarity
temp_df.rename(columns={
    'YEAR': 'Year',
    'ANNUAL': 'Annual_Temp',
    'JAN-FEB': 'Temp_JanFeb',
    'MAR-MAY': 'Temp_MarMay',
    'JUN-SEP': 'Temp_JunSep',
    'OCT-DEC': 'Temp_OctDec'
}, inplace=True)

# ✅ Merge on 'Year'
merged_df = pd.merge(daily_df, temp_df, on='Year', how='left')

# ✅ Save final merged dataset
merged_df.to_csv("final_combined_data_with_temperature.csv", index=False)
print("✅ Final dataset with temperature merged saved as: final_combined_data_with_temperature.csv")
```

```python
import pandas as pd

# Load the already-merged file
df = pd.read_csv("final_combined_data_with_temperature.csv", parse_dates=['Date'], dayfirst=True)

# Function to map Month to the correct seasonal temperature
def get_temp_from_month(row):
    month = row['Month_rain']
    if month in [1, 2]:
        return row['Temp_JanFeb']
    elif month in [3, 4, 5]:
        return row['Temp_MarMay']
    elif month in [6, 7, 8, 9]:
        return row['Temp_JunSep']
    elif month in [10, 11, 12]:
        return row['Temp_OctDec']
    else:
        return None  # fallback in case of bad data

# Apply the mapping
df['Temp_Seasonal'] = df.apply(get_temp_from_month, axis=1)

# Drop the seasonal columns if you no longer need them
df.drop(columns=['Temp_JanFeb', 'Temp_MarMay', 'Temp_JunSep', 'Temp_OctDec'], inplace=True)

# Save updated file
df.to_csv("final_combined_data_with_single_temperature.csv", index=False)
print("✅ Combined seasonal temperatures into single column 'Temp_Seasonal'. File saved.")
```

```python
import pandas as pd
import numpy as np

# Load your combined rainfall + soil + temperature data
df = pd.read_csv("temp_soil_rain_combined.csv", parse_dates=['Date'], dayfirst=True)

# Sort by State, District, and Date (important for lagged and rolling features)
df.sort_values(by=['State', 'District', 'Date'], inplace=True)

# Group by State and District to calculate region-specific features
group = df.groupby(['State', 'District'])

# ✅ Lag features
df['Rainfall_lag_1d'] = group['Avg_rainfall'].shift(1)
df['Rainfall_lag_3d'] = group['Avg_rainfall'].shift(3)
df['Rainfall_lag_7d'] = group['Avg_rainfall'].shift(7)

# ✅ Rolling (7-day) rainfall features
df['Rainfall_7d_sum'] = group['Avg_rainfall'].rolling(window=7, min_periods=1).sum().reset_index(level=[0, 1], drop=True)
df['Rainfall_7d_avg'] = group['Avg_rainfall'].rolling(window=7, min_periods=1).mean().reset_index(level=[0, 1], drop=True)

# ✅ 7-day average soil moisture
df['SoilMoisture_7d_avg'] = group['Avg_smlvl_at15cm'].rolling(window=7, min_periods=1).mean().reset_index(level=[0, 1], drop=True)

# ✅ Time-based features
df['day_of_year'] = df['Date'].dt.dayofyear
df['week_of_year'] = df['Date'].dt.isocalendar().week
df['month'] = df['Date'].dt.month

# ✅ Cyclic month encoding
df['month_sin'] = np.sin(2 * np.pi * df['month'] / 12)
df['month_cos'] = np.cos(2 * np.pi * df['month'] / 12)

# Optional: drop the 'month' column if not needed
df.drop(columns=['month'], inplace=True)

# ✅ Save to new file
df.to_csv("final_dataset_with_features.csv", index=False)
print("✅ Feature-rich dataset saved as 'final_dataset_with_features.csv'")
```
